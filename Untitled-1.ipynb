```python
import pandas as pd

# 1. Leer el CSV indicando el separador correcto
df_pacientes = pd.read_csv("Genotype Matrix.csv", sep=";")

# 2. Empezar a crear una función que evalúe a un paciente (fila)
def obtener_diplotipo_cyp2d6(fila):
    alelos = []
    
    # Comprobamos cada columna. Si tiene variante, añadimos el alelo correspondiente.
    # Nota: Aquí asumo que T/T es normal para *3. ¡Deberás confirmar esto con tu teoría!
    if fila['CYP2D6*4'] == 'C/T':
        alelos.append('*4')
    elif fila['CYP2D6*4'] == 'T/T': 
        alelos.extend(['*4', '*4']) # Homocigoto variante
        
    # ... (habría que hacer lo mismo para *3, *17 y *41) ...
    
    # Aplicar la REGLA DE ORO: Si no hemos encontrado variantes, rellenamos con *1
    while len(alelos) < 2:
        alelos.append('*1')
        
    # Devolvemos el diplotipo formateado, ej: "*1/*4"
    return f"{alelos[0]}/{alelos[1]}"

# Aplicamos la función a nuestra tabla de pacientes
df_pacientes['Diplotipo_CYP2D6'] = df_pacientes.apply(obtener_diplotipo_cyp2d6, axis=1)

print(df_pacientes[['Sample/Assay', 'Diplotipo_CYP2D6']].head())
```
